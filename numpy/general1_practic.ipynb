```python
import numpy as np 
```

```python
arr1 = np.array([10, 20, 30, 40, 50])
arr2= np.array([1, 2, 3, 4, 5])
print("sum            :",arr1+arr2)
print("subtraction    :",arr1-arr2)
print("multiplication :",arr1*arr2)
print("division       :",arr1/arr2)
print("power of       :",arr1**2)
print("power of       :",arr2**2)

```

```python
arr3=np.array([75, 80, 65, 90, 85, 70, 95])
print("the toatal marks is  :",np.sum(arr3))
print("the average is       :",np.mean(arr3))
print("the maximum value is :",np.max(arr3))
print("the minimum value is :",np.min(arr3))
print("the standard derivation is ",np.std(arr3))
```

```python
arr4=np.array([12, 45, 67, 23, 89, 34, 90, 56, 10, 78])
print("elements greater then 50 :",arr4>50)
print("elements greater then 50 :",arr4<30)
print("elements greater then 50 :",arr4%2==0)
print("between 30 to 70         :",(arr4>30) & (arr4<70))
```

```python
var= np.arange(1, 13)
arr5=var.reshape(3,4)
arr6=var.reshape(2,6)
arr7=arr5.flatten()
print(arr5)
print(arr6)
print(arr7)
```

```python
a = np.array([10, 20, 30])
b = np.array([40, 50, 60])
arr8= np.concatenate([a,b])
print(arr8)
arr9= np.vstack([a,b])
print(arr9)
arr10= np.hstack([a,b])
print(arr10)
print(arr8.shape)
print(arr9.shape)
print(arr10.shape)
```

```python


arr11 = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
print(np.sum(arr11))
print("sum of Rows ",np.sum(arr11, axis=1))
print()
print("sum of column",np.sum(arr11, axis=0))
print("maximum ",np.max(arr11))
print("minimum ",np.min(arr11))
print("average of each row ",np.mean(arr11, axis=1))
```

```python
arr12 = np.array([45, 12, 89, 23, 67, 12, 90, 34])
print("ascending form       :", np.sort(arr12))
print("decending form       :", np.sort(arr12)[::-1])
print("smallest value       :",np.min(arr12))
print("largest  value       :",np.max(arr12))
print("the original form is :",arr12)
```

```python
marks = np.array([
    [78, 85, 90],
    [65, 70, 72],
    [92, 88, 95],
    [56, 60, 68]
])
student1_total=np.sum(marks[0])
student2_total=np.sum(marks[1])
student3_total=np.sum(marks[2])
student4_total=np.sum(marks[3])
print(f"the total marks of each studnet is : {student1_total}")
print(f"the total marks of each studnet is : {student2_total}")
print(f"the total marks of each studnet is : {student3_total}")
print(f"the total marks of each studnet is : {student4_total}")
student1_average=np.mean(marks[0])
student2_average=np.mean(marks[1])
student3_average=np.mean(marks[2])
student4_average=np.mean(marks[3])
print()
print(f"the average  of each studnet is : {student1_average}")
print(f"the average  of each studnet is : {student2_average}")
print(f"the average  of each studnet is : {student3_average}")
print(f"the average  of each studnet is : {student4_average}\n")
subject_highest= np.max(marks,axis=0)
print("highest marks is each sunject is  ",subject_highest)
print()
subject_average= np.mean(marks,axis=0)
print("average marks is each sunject is  ",subject_average)
print()
all_student_average= np.mean(marks,axis=1)
print(f"student average marks is : {all_student_average}")
print("the student marks is decending form :",np.sort(all_student_average[::-1]))

```
