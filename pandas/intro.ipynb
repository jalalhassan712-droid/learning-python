```python
marks = pd.Series([65, 85, 94,88], index= ["ali", "ahmad","khan","zaman"])
print(marks)
print("ali marks is :",marks["ali"])
print("ahmad marks is :",marks["ahmad"])
print("average marks is :", marks.mean())
print("highest marks is :", marks.max())
print("lowest marks is :", marks.min())
ubdate_marks = marks + 5
print(ubdate_marks)
```

```python
sales = pd.Series([120000,85000 , 45000, 15000, 10000, 5000], index=["laptop","mobile", "tablets", "headphones","keyboard", "mouse"])
print(sales)
print("laptop salse is : ",sales["laptop"])
print("total sales is :",sales.sum())
print("averag sales is :",sales.mean())
high = sales.max()
print("highest sales is :",high )
low = sales.min()
print("lowest sales is :", low)
difference = high-low
print("differnce between highest and lowest is ", difference)
new_sales= sales * 1.20
print("ubdated sales is ", new_sales)
print("sales which is greater then 100000")
print(new_sales[new_sales > 100000])
```
