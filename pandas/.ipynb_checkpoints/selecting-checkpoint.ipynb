```python
df.loc[0]                # First row (by label)
```

# Select Specific Rows and Columns

```python
df.iloc[1,2]
```

```python
df.loc[1, "city"]     #also use column name 
```

# slicing 

```python
df.iloc[0:3]
```

```python
df.loc[4:6, ["team1", "team2"]]   # Rows 0 to 2, selected columns
```

```python
df.iloc[0:2, 0:2]              # Rows and cols by index position
```

```python
df.iloc[:,2]
```
